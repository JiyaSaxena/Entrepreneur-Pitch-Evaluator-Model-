```python

import os
import pandas as pd
import numpy as np

BASE = "/content/Entrepreneur-Pitch-Evaluator-Model-"

startup_df = pd.read_csv(
    os.path.join(BASE, "startup_test.csv"),
    low_memory=False
)

sharktank_df = pd.read_csv(
    os.path.join(BASE, "Shark Tank India.csv"),
    low_memory=False
)

# Prepare startup descriptions
startup_df["startup_description"] = (
    startup_df["startup_description"]
    .fillna("")
    .astype(str)
)

startup_df["clean_description"] = (
    startup_df["clean_description"]
    .fillna("")
    .astype(str)
)

startup_df["pitch_text"] = startup_df["clean_description"].where(
    startup_df["clean_description"].str.strip().ne(""),
    startup_df["startup_description"]
)

startup_df = startup_df[
    startup_df["pitch_text"].str.strip().ne("")
].reset_index(drop=True)

# Prepare Shark Tank India descriptions
sharktank_df["Business Description"] = (
    sharktank_df["Business Description"]
    .fillna("")
    .astype(str)
)

sharktank_df = sharktank_df[
    sharktank_df["Business Description"].str.strip().ne("")
].reset_index(drop=True)

print("Startup records:", len(startup_df))
print("Shark Tank India records:", len(sharktank_df))

print("\nExample startup:")
print(startup_df.loc[0, "startup_name"])
print(startup_df.loc[0, "pitch_text"][:500])

print("\nExample Shark Tank India pitch:")
print(sharktank_df.loc[0, "Startup Name"])
print(sharktank_df.loc[0, "Business Description"][:500])
```

```python

import torch
from sentence_transformers import SentenceTransformer

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", device)

retrieval_df = pd.concat(
    [
        pd.DataFrame({
            "startup_name": startup_df["startup_name"].astype(str),
            "industry": startup_df["industry"].fillna("Unknown").astype(str),
            "pitch_text": startup_df["pitch_text"],
            "source": "Startup Dataset"
        }),
        pd.DataFrame({
            "startup_name": sharktank_df["Startup Name"].astype(str),
            "industry": sharktank_df["Industry"].fillna("Unknown").astype(str),
            "pitch_text": sharktank_df["Business Description"],
            "source": "Shark Tank India"
        })
    ],
    ignore_index=True
)

retrieval_df = retrieval_df[
    retrieval_df["pitch_text"].str.strip().ne("")
].drop_duplicates(
    subset=["startup_name", "pitch_text"]
).reset_index(drop=True)

retrieval_model = SentenceTransformer(
    "all-MiniLM-L6-v2",
    device=device
)

pitch_embeddings = retrieval_model.encode(
    retrieval_df["pitch_text"].tolist(),
    batch_size=64,
    show_progress_bar=True,
    convert_to_numpy=True,
    normalize_embeddings=True
)

print("Records indexed:", len(retrieval_df))
print("Embedding matrix shape:", pitch_embeddings.shape)
```

```python

from sentence_transformers import util

def retrieve_similar_pitches(pitch, top_k=10):
    query_embedding = retrieval_model.encode(
        pitch,
        convert_to_tensor=True,
        normalize_embeddings=True
    )

    scores = util.cos_sim(
        query_embedding,
        torch.tensor(pitch_embeddings, device=query_embedding.device)
    )[0]

    k = min(top_k, len(retrieval_df))
    values, indices = torch.topk(scores, k=k)

    results = retrieval_df.iloc[
        indices.cpu().numpy()
    ].copy()

    results["similarity"] = values.cpu().numpy().round(4)

    return results[
        ["startup_name", "industry", "source", "similarity", "pitch_text"]
    ]

test_pitch = """
We sell healthy ready-to-eat Indian snacks made from millets.
Our company targets urban consumers looking for convenient,
nutritious food products through online sales and retail stores.
"""

retrieve_similar_pitches(test_pitch, top_k=5)
```

```python

from sentence_transformers import CrossEncoder
import pandas as pd

reranker = CrossEncoder(
    "cross-encoder/ms-marco-MiniLM-L-6-v2",
    device=device
)

def rerank_similar_pitches(pitch, top_k=5, candidates=20):
    # First retrieve a broader set using the existing model
    initial = retrieve_similar_pitches(pitch, top_k=candidates)

    # Cross-Encoder jointly scores the pitch and each candidate
    pairs = [
        (pitch, row["pitch_text"])
        for _, row in initial.iterrows()
    ]

    scores = reranker.predict(pairs, show_progress_bar=False)

    initial = initial.copy()
    initial["reranker_score"] = scores

    return initial.sort_values(
        "reranker_score", ascending=False
    ).head(top_k).reset_index(drop=True)

test_pitch = """
We manufacture healthy millet-based snacks for urban consumers.
We plan to sell through our website, supermarkets and food distributors.
"""

reranked_results = rerank_similar_pitches(test_pitch, top_k=5)
display(
    reranked_results[
        ["startup_name", "industry", "source", "similarity", "reranker_score"]
    ]
)
```

```python

from transformers import pipeline
import torch

classifier_device = 0 if torch.cuda.is_available() else -1

industry_classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli",
    device=classifier_device
)

INDUSTRIES = [
    "FoodTech and food products",
    "Healthcare and medical technology",
    "Financial technology and fintech",
    "Education technology",
    "Artificial intelligence and software",
    "E-commerce and retail",
    "Clean energy and sustainability",
    "Agriculture technology",
    "Consumer products and lifestyle",
    "Logistics and transportation"
]

def classify_industry(pitch):
    result = industry_classifier(
        pitch,
        candidate_labels=INDUSTRIES,
        multi_label=False
    )

    return pd.DataFrame({
        "industry": result["labels"],
        "confidence": result["scores"]
    })

display(classify_industry(test_pitch).head(5))
```

```python

from transformers import pipeline

sentiment_device = 0 if torch.cuda.is_available() else -1

finbert = pipeline(
    "text-classification",
    model="ProsusAI/finbert",
    device=sentiment_device
)

def analyze_news_sentiment(text):
    result = finbert(
        str(text)[:2000],
        truncation=True
    )[0]

    return {
        "sentiment": result["label"],
        "confidence": round(float(result["score"]), 4)
    }

print(analyze_news_sentiment(
    "Investors increase funding for promising food technology startups."
))
```

```python

import feedparser
from urllib.parse import quote

def get_live_news(industry, max_results=5):
    query = quote(f"{industry} startup funding market")
    url = (
        "https://news.google.com/rss/search"
        f"?q={query}&hl=en-IN&gl=IN&ceid=IN:en"
    )

    feed = feedparser.parse(url)
    articles = []

    for entry in feed.entries[:max_results]:
        title = entry.get("title", "Untitled")
        sentiment = analyze_news_sentiment(title)

        articles.append({
            "title": title,
            "link": entry.get("link", ""),
            "published": entry.get("published", "Date unavailable"),
            "sentiment": sentiment["sentiment"],
            "sentiment_confidence": sentiment["confidence"]
        })

    return pd.DataFrame(articles)

display(get_live_news("FoodTech", max_results=5))
```

```python

def analyze_pitch(pitch, location="India"):
    industry_results = classify_industry(pitch)
    predicted_industry = industry_results.iloc[0]["industry"]

    similar = rerank_similar_pitches(
        pitch,
        top_k=5,
        candidates=20
    )

    news = get_live_news(predicted_industry, max_results=5)

    return {
        "predicted_industry": predicted_industry,
        "industry_scores": industry_results.head(5),
        "similar_startups": similar,
        "market_news": news,
        "location": location
    }

demo_pitch = """
Our startup produces affordable, healthy millet snacks for working
professionals. We sell online and want to expand into supermarkets
and local food distribution networks.
"""

analysis_result = analyze_pitch(demo_pitch, "Mumbai, India")

print("Predicted industry:", analysis_result["predicted_industry"])
print("\nTop similar startups:")
display(analysis_result["similar_startups"])
print("\nMarket news:")
display(analysis_result["market_news"])
```

```python

import gradio as gr
from urllib.parse import quote_plus
import pandas as pd

def make_pitch_report(pitch, location):
    result = analyze_pitch(pitch, location)

    industry = result["predicted_industry"]
    industry_scores = result["industry_scores"]
    similar = result["similar_startups"]
    news = result["market_news"]

    report = f"""
# 🚀 Startup Pitch Evaluation

## 1. Pitch overview

**Detected industry:** {industry}

**Location:** {location}

**Submitted pitch:**

{pitch}

## 2. Industry classification

"""
    for _, row in industry_scores.iterrows():
        report += (
            f"- **{row['industry']}** — "
            f"{row['confidence']:.1%} model confidence\n"
        )

    report += "\n## 3. Similar startup pitches\n\n"

    for _, row in similar.iterrows():
        report += (
            f"### {row['startup_name']}\n"
            f"- Industry: {row['industry']}\n"
            f"- Source: {row['source']}\n"
            f"- Semantic similarity: {row['similarity']:.3f}\n"
            f"- Reranker score: {row['reranker_score']:.3f}\n"
            f"- Description: {str(row['pitch_text'])[:350]}\n\n"
        )

    report += "## 4. Industry news and sentiment\n\n"

    if news.empty:
        report += "No news articles were returned. Try again later.\n"
    else:
        for _, row in news.iterrows():
            report += (
                f"- [{row['title']}]({row['link']})\n"
                f"  - Published: {row['published']}\n"
                f"  - FinBERT sentiment: **{row['sentiment']}** "
                f"(confidence: {row['sentiment_confidence']:.2f})\n\n"
            )

    supplier_query = quote_plus(
        f"{industry} suppliers distributors {location}"
    )
    market_query = quote_plus(
        f"{industry} market competitors {location}"
    )
    funding_query = quote_plus(
        f"{industry} startup funding investors India"
    )
    map_query = quote_plus(
        f"{industry} suppliers distributors near {location}"
    )

    report += f"""
## 5. Business research links

These are search links to help you research opportunities. They are
not verified supplier recommendations.

- [Find suppliers and distributors](https://www.google.com/search?q={supplier_query})
- [Research competitors and the market](https://www.google.com/search?q={market_query})
- [Find startup investors and funding information](https://www.google.com/search?q={funding_query})
- [Locate relevant businesses on Google Maps](https://www.google.com/maps/search/{map_query})

## 6. Preliminary recommendations

- Validate demand with potential customers before scaling.
- Compare competitor pricing, distribution channels and product differentiation.
- Prepare realistic unit economics, gross margins and customer acquisition costs.
- Verify supplier credentials, product quality and delivery terms independently.

**Important:** Similarity, industry confidence and news sentiment are model outputs.
They do not establish business viability or guarantee investment.
"""

    return report, result


def run_pitch_analysis(pitch, location):
    if not pitch or not pitch.strip():
        return "Please enter a startup pitch.", None, "", location

    try:
        report, result = make_pitch_report(pitch.strip(), location.strip() or "India")
        return report, result, pitch, location
    except Exception as e:
        return f"Analysis failed: {type(e).__name__}: {e}", None, pitch, location
```

```python

def pitch_chat(message, history, analysis, pitch, location):
    if not message or not message.strip():
        return "Ask me something about your startup pitch."

    if not analysis:
        return "Please submit a pitch using **Analyze My Pitch** first."

    q = message.lower()
    industry = analysis["predicted_industry"]
    similar = analysis["similar_startups"]
    news = analysis["market_news"]

    if any(word in q for word in ["industry", "sector", "category"]):
        scores = analysis["industry_scores"]
        lines = [
            f"Your pitch is most closely classified as **{industry}**.",
            "",
            "Other industry scores:"
        ]
        for _, row in scores.head(5).iterrows():
            lines.append(
                f"- {row['industry']}: {row['confidence']:.1%}"
            )
        return "\n".join(lines)

    if any(word in q for word in [
        "similar", "competitor", "competition", "startup"
    ]):
        lines = ["Here are the closest retrieved startup pitches:", ""]
        for _, row in similar.head(5).iterrows():
            lines.append(
                f"- **{row['startup_name']}** "
                f"({row['industry']}; {row['source']}) — "
                f"similarity {row['similarity']:.3f}"
            )
        lines += [
            "",
            "These are similarity matches, not necessarily direct competitors."
        ]
        return "\n".join(lines)

    if any(word in q for word in ["news", "market", "sentiment", "trend"]):
        if news.empty:
            return "No news was retrieved. Please retry the analysis later."

        lines = ["Recent retrieved industry headlines:", ""]
        for _, row in news.head(5).iterrows():
            lines.append(
                f"- [{row['title']}]({row['link']}) — "
                f"{row['sentiment']} sentiment "
                f"(confidence {row['sentiment_confidence']:.2f})"
            )
        lines += [
            "",
            "Sentiment is inferred from headlines and may not reflect the full article."
        ]
        return "\n".join(lines)

    if any(word in q for word in [
        "supplier", "distributor", "vendor", "location", "where"
    ]):
        query = quote_plus(f"{industry} suppliers distributors {location}")
        maps = quote_plus(f"{industry} suppliers near {location}")
        return (
            f"For your **{industry}** pitch, start researching suppliers "
            f"and distributors around **{location}**.\n\n"
            f"- [Search suppliers and distributors](https://www.google.com/search?q={query})\n"
            f"- [Search nearby businesses on Maps](https://www.google.com/maps/search/{maps})\n\n"
            "These are research links, not verified supplier listings."
        )

    if any(word in q for word in [
        "improve", "feedback", "weakness", "recommend", "advice", "score"
    ]):
        return (
            f"Your pitch is classified as **{industry}**. To strengthen it, "
            "explain the customer problem, target market, unique advantage, "
            "pricing, competitors, distribution strategy and evidence of demand. "
            "Include revenue, costs, gross margin and funding requirements where "
            "available. I cannot give a reliable investment score without "
            "validated financial and market evidence."
        )

    return (
        f"Your current pitch analysis focuses on **{industry}**.\n\n"
        "Try asking me about:\n"
        "- Industry classification\n"
        "- Similar startups and competitors\n"
        "- Live market news and sentiment\n"
        "- Suppliers and distributors near your location\n"
        "- Ways to improve your pitch"
    )
```

```python

with gr.Blocks(title="Startup Pitch Evaluator AI") as app:
    gr.Markdown(
        """
        # 🚀 Startup Pitch Evaluator AI
        ### Deep-learning-powered startup research assistant

        Enter your business idea to classify its industry, find similar
        startup pitches, examine market headlines and explore business links.
        """
    )

    pitch_input = gr.Textbox(
        label="Describe your startup pitch",
        placeholder=(
            "Example: We produce affordable millet snacks for urban "
            "consumers and sell through online platforms and supermarkets."
        ),
        lines=5
    )

    location_input = gr.Textbox(
        label="Target location",
        value="Mumbai, India"
    )

    analyze_button = gr.Button("Analyze My Pitch", variant="primary")
    report_output = gr.Markdown()

    analysis_state = gr.State(None)
    pitch_state = gr.State("")
    location_state = gr.State("India")

    analyze_button.click(
        fn=run_pitch_analysis,
        inputs=[pitch_input, location_input],
        outputs=[
            report_output,
            analysis_state,
            pitch_state,
            location_state
        ]
    )

    gr.Markdown("---")
    gr.Markdown("## 💬 Ask questions about your pitch")

    gr.ChatInterface(
        fn=pitch_chat,
        additional_inputs=[
            analysis_state,
            pitch_state,
            location_state
        ],
        title="Entrepreneur Assistant",
        description=(
            "Analyze your pitch first, then ask about competitors, "
            "industry trends, news or suppliers."
        )
    )

app.launch(share=True)
```
